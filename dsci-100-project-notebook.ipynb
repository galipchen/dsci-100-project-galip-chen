The plan for this project is to use K-nearest neighbors classification to predict a team's tournament finish using information about their seed and regional win percentage. This method is appropriate for the purposes of this analysis because the variable I am trying to predict is categorical and the predictors are numerical. 

One of the main weaknesses of using a KNN model is that stronger teams with higher seeds and win percentages can easily lose to weaker teams due to extenuating circumstances like injuries or teams simply having a bad game, especially for single-elimination tournaments. These are factors which seed and regular-season win percentage can’t capture. Also, the model may not perform as well in the case of imbalanced classes since earlier finishes are more common then later finishes, so the model may rarely predict the later finishes.

I plan to split the data into 75% training set and 25% testing set, stratified by tourney_finish. I will also set a seed for reproducibility. In addition, I will be using 5 fold cross-validation on the training set to tune for the number of nearest neigbours (K), selecting the K with the highest predicting accuracy. Lastly, I will be using a recipe that standardizes the predictors so that all predictors have equal contributions in affecting the final prediction.

I will evaluate the final model on the test set using accuracy and a confusion matrix. Accuracy summarizes overall performance of the model which is appropriate in answering the question of whether the three predictors can be used to accurately predict tournament finish, while the confusion matrix shows which finishes are misclassified.

```r
### EDIT THIS CODE CELL ###

# Load the tidymodels package for building the classification model
library(tidymodels)

# Set seed for reproducibility
set.seed(40791741)


# Split the data (75% train / 25% test), stratified by tourney_finish
split_data <- initial_split(march_madness_tidy, prop = 0.75, strata = tourney_finish)
train_data <- training(split_data)
test_data  <- testing(split_data)

# Create a recipe that standardizes the predictors 
knn_recipe <- recipe(tourney_finish ~ seed + reg_wins_pct, data = train_data) |>
  step_scale(all_predictors()) |>
  step_center(all_predictors())

# Model specification with K to be tuned
knn_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>
  set_engine("kknn") |>
  set_mode("classification")

# 5-fold-cross-validation on the training set
knn_vfold <- vfold_cv(train_data, v = 5, strata = tourney_finish)

# K values to try
k_vals <- tibble(neighbors = seq(from = 1, to = 50, by = 2))

# Use cross-validation to tune for K
knn_results <- workflow() |>
  add_recipe(knn_recipe) |>
  add_model(knn_spec) |>
  tune_grid(resamples = knn_vfold, grid = k_vals) |>
  collect_metrics()

# Keep only accuracy results
accuracies <- knn_results |>
  filter(.metric == "accuracy")

# Select the K with the highest cross-validation accuracy
tuned_k <- accuracies |>
  arrange(desc(mean)) |>
  slice(1) |>
  pull(neighbors)

tuned_k

# Fit the final model on the full training set using best K
final_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tuned_k) |>
  set_engine("kknn") |>
  set_mode("classification")

final_fit <- workflow() |>
  add_recipe(knn_recipe) |>
  add_model(final_spec) |>
  fit(data = train_data)
```
